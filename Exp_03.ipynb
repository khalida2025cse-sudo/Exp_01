```python
import pandas as pd
import numpy as np

data = {
    "CustomerID": [101,102,103,104,105,106,107,108,109,109,110],

    "Age_Group": [
        "18-25","26-30","31-35","18-25","36+",
        "26-30","31-35","36+","18-25","18-25","26-30"
    ],

    "Rating(1-5)": [5,4,6,2,-2,3,5,8,4,4,1],

    "Hotel": [
        "Taj", "Marriott", "Hilton", "Taj", "Oberoi",
        "Marriot", "Hilton", "Taj", "Oberoi", "Oberoi", "Hilton"
    ],

    "FoodPreference": [
        "veg", "Non-Veg", "Vegetarian", "Veg", "non-Veg",
        "Non-Veg", "Veg", "Vegetarian", "veg", "Non-Veg", "Vegetarian"
    ],

    "Bill": [
        2500, 3200, 1800, 4500, 2100,
        -500, 3900, 5200, 2750, 2750, 6100
    ],

    "NoOfPax": [
        2,3,4,2,5,2,-1,3,4,4,2
    ],

    "EstimatedSalary": [
        45000,62000,75000,38000,90000,
        -5000,68000,110000,-25000,-25000,82000
    ]
}

df = pd.DataFrame(data)

print("Original Dataset:")
print(df)


df.drop_duplicates(inplace=True)


df.reset_index(drop=True, inplace=True)


df.loc[df["Bill"] < 0, "Bill"] = np.nan

df.loc[df["EstimatedSalary"] < 0, "EstimatedSalary"] = np.nan

df.loc[df["NoOfPax"] < 1, "NoOfPax"] = np.nan

df.loc[
    (df["Rating(1-5)"] < 1) | (df["Rating(1-5)"] > 5),
    "Rating(1-5)"
] = np.nan


df["Hotel"] = df["Hotel"].replace("Marriot", "Marriott")


df["FoodPreference"] = df["FoodPreference"].replace(
    ["veg", "Vegetarian"], "Veg"
)

df["FoodPreference"] = df["FoodPreference"].replace(
    "non-Veg", "Non-Veg"
)


df["Bill"] = df["Bill"].fillna(df["Bill"].mean())

df["EstimatedSalary"] = df["EstimatedSalary"].fillna(
    df["EstimatedSalary"].mean()
)

df["NoOfPax"] = df["NoOfPax"].fillna(
    df["NoOfPax"].median()
)

df["Rating(1-5)"] = df["Rating(1-5)"].fillna(
    df["Rating(1-5)"].median()
)

print("\nCleaned Dataset:")
print(df)
```
