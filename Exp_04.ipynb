```python
import numpy as np
import pandas as pd

data = {
    "Country": [
        "India", "USA", "UK", "India", "Canada",
        "USA", "India", np.nan, "UK", "Canada"
    ],

    "Age": [
        22, 25, np.nan, 31, 28,
        35, 24, 40, np.nan, 30
    ],

    "Salary": [
        35000, 50000, 45000, np.nan, 60000,
        55000, 42000, 70000, 48000, np.nan
    ],

    "Purchased": [
        "Yes", "No", "Yes", "No", "Yes",
        "No", "Yes", "Yes", "No", "Yes"
    ]
}

df = pd.DataFrame(data)

print("Original Dataset:")
display(df)

print("Dataset Information:")
df.info()

print("Mode of Country:")
print(df["Country"].mode())

# Fill missing Country with Mode
df["Country"] = df["Country"].fillna(df["Country"].mode()[0])

# Fill missing Age with Median
df["Age"] = df["Age"].fillna(df["Age"].median())

# Fill missing Salary with Mean
df["Salary"] = df["Salary"].fillna(round(df["Salary"].mean()))

print("After Data Preprocessing:")
display(df)
```
