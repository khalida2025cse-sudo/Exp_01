```python
import pandas as pd
import matplotlib.pyplot as plt

data = {
    'Year': list(range(2010, 2021)),
    'Job Postings': [100, 250, 380, 520, 700, 950, 1250, 1550, 1900, 2300, 2800]
}

df = pd.DataFrame(data)

plt.plot(df['Year'], df['Job Postings'], marker='o')
plt.title('Trend of Data Science Job Postings')
plt.xlabel('Year')
plt.ylabel('Number of Job Postings')
plt.show()

```

```python

import pandas as pd
import matplotlib.pyplot as plt

data = {
    'Role': ['Data Analyst', 'Data Engineer', 'Data Scientist',
             'ML Engineer', 'Data Analyst', 'Data Scientist',
             'Data Engineer', 'ML Engineer', 'Data Analyst', 'Data Scientist']
}

df = pd.DataFrame(data)

role_count = df['Role'].value_counts()

print(role_count)

role_count.plot(kind='bar')
plt.title('Distribution of Data Science Roles')
plt.xlabel('Job Role')
plt.ylabel('Number of Jobs')
plt.xticks(rotation=45)
plt.show()
```
